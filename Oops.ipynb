Q1: What is Object-Oriented Programming (OOP)?

Answer:
Object-Oriented Programming (OOP) is a programming paradigm based on the concept of “objects.” These objects contain both data (attributes) and functions (methods) that operate on that data.
The main idea behind OOP is to model real-world entities into classes and objects, making the code more organized, reusable, and easier to maintain.
OOP provides several key principles — encapsulation, inheritance, abstraction, and polymorphism — that together promote cleaner and more modular program design.
Python is a fully object-oriented language where everything is treated as an object, including numbers, strings, and functions.

Q2: What is a class in OOP?

Answer:
A class in OOP is a blueprint or template that defines the structure and behavior (data and methods) that its objects will have.
It acts like a user-defined data type that describes how to create objects with specific properties and actions.
For example, a Car class may define attributes like color and speed, and methods like start() and stop().
Creating a class improves code modularity, since all related data and logic are grouped in one place.

Q3: What is an object in OOP?

Answer:
An object is an instance of a class. It represents a specific entity created from the class blueprint.
When a class is defined, no memory is allocated; memory is only assigned when an object is created.
Each object can have its own unique attribute values, even though they belong to the same class.
For example, if Car is a class, then car1 = Car("Red", 120) and car2 = Car("Blue", 150) are two different objects of that class.

Q4: What is the difference between abstraction and encapsulation?

Answer:
Both are fundamental OOP concepts, but they serve different purposes:

Abstraction: Focuses on hiding the complex implementation details and exposing only the necessary functionality. It helps simplify interactions with the system. For example, when we use a car, we only press the accelerator or brake without knowing how the engine works.

Encapsulation: Refers to wrapping data and methods into a single unit (class) and controlling access using access modifiers. It ensures data protection and prevents accidental modification of internal details.
In simple terms — Abstraction hides complexity, and Encapsulation protects data.

Q5: What are dunder methods in Python?

Answer:
Dunder methods (short for double underscore methods) are special predefined methods in Python that begin and end with double underscores (__method__).
They allow developers to customize the behavior of built-in operations such as object initialization, representation, comparison, and arithmetic operations.
For example:

__init__ initializes an object.

__str__ returns a user-friendly string representation.

__add__ allows operator overloading for addition.
These methods make classes behave more like built-in Python objects and enable operator overloading and custom object behavior.

Q6: Explain the concept of inheritance in OOP.

Answer:
Inheritance is a mechanism in OOP that allows one class (child or subclass) to acquire the properties and methods of another class (parent or superclass).
It promotes code reusability and enables building a class hierarchy.
In Python, inheritance is achieved by passing the parent class as a parameter to the child class.
For example:

class Parent:
    def speak(self):
        print("Parent speaking")

class Child(Parent):
    def speak(self):
        print("Child speaking")


Here, the Child class inherits from Parent, and we can override or extend parent behavior.
Python also supports multiple and multilevel inheritance.

Q7: What is polymorphism in OOP?

Answer:
Polymorphism means “many forms.” It allows the same method or function to behave differently depending on the object calling it.
In Python, polymorphism is achieved through method overriding and duck typing.

Q8: How is encapsulation achieved in Python?

Answer:
Encapsulation in Python is achieved by restricting direct access to class variables using access modifiers:

Public attributes: can be accessed anywhere.

Protected attributes: start with a single underscore _attribute.

Private attributes: start with double underscores __attribute.
Getter and setter methods (or property decorators) are then used to control access to these variables.
Encapsulation improves security and maintains data integrity.

Q9: What is a constructor in Python?

Answer:
A constructor is a special method named __init__() that is automatically called when an object is created.
It initializes object attributes and sets up the initial state.

Q10: What are class and static methods in Python?

Answer:
Python provides two types of methods that don’t depend on object instances:

Class Method: Declared using @classmethod, takes cls as the first argument, and can modify class-level attributes shared among all instances.

Static Method: Declared using @staticmethod, doesn’t take self or cls. It acts like a normal function inside the class, used for logical grouping.

Q11: What is method overloading in Python?

Answer:
Method overloading allows multiple methods in the same class with the same name but different numbers or types of parameters.
However, Python does not support traditional method overloading like Java or C++.
In Python, the last defined method with the same name overrides all previous ones.
To achieve similar behavior, we can use default arguments or variable-length arguments (*args, **kwargs) to handle different numbers of parameters.

Q12: What is method overriding in OOP?

Answer:
Method overriding occurs when a subclass provides a new implementation for a method that already exists in its parent class.
This allows child classes to customize or completely replace inherited behavior.
The overridden method in the child class must have the same name and parameters as in the parent class.

Q13: What is a property decorator in Python?

Answer:
The @property decorator in Python is used to make class methods act like attributes.
It allows defining getter, setter, and deleter functions for encapsulated data.
This improves readability and provides control over how attributes are accessed or modified.

Q14: Why is polymorphism important in OOP?

Answer:
Polymorphism increases flexibility and scalability in OOP.
It allows objects of different classes to be treated under a common interface — making code more reusable and maintainable.
For example, we can loop through a list of objects of different classes and call the same method (fly(), speak(), etc.), and each object will respond differently.
Polymorphism makes it easier to extend programs without modifying existing code — a key OOP advantage.

Q15: What is an abstract class in Python?

Answer:
An abstract class defines a common structure for its subclasses but cannot be instantiated directly.
It can contain one or more abstract methods, which must be implemented by subclasses.
Python uses the abc (Abstract Base Class) module to create abstract classes.

Object-Oriented Programming provides several benefits:

Q17: What is the difference between a class variable and an instance variable?

Answer:

Class Variable:
Shared among all instances of the class. It belongs to the class itself. Changing it affects all objects.

Instance Variable:
Belongs to a specific object (defined using self). Each object has its own copy.

18: What is multiple inheritance in Python?

Answer:
Multiple inheritance allows a class to inherit from more than one parent class.
This lets the child class access attributes and methods from all parents.
However, it can lead to ambiguity when methods with the same name exist in multiple parents — Python resolves this using the Method Resolution Order (MRO).

Q19: Explain the purpose of __str__ and __repr__ methods in Python.

Answer:
Both methods define how an object is represented as a string:
__str__ → Returns a readable, user-friendly representation (used by print() or str()).
__repr__ → Returns a developer-friendly string used for debugging and logging.
If __str__ is not defined, __repr__ is used as a fallback.

Q20: What is the significance of the super() function in Python?

Answer:
super() is used to call methods from the parent (superclass) in a subclass.
It is commonly used inside constructors to initialize inherited attributes or extend parent behavior.
It ensures that the parent’s logic executes before or along with the child’s.

Q21: What is the significance of the __del__ method in Python?

Answer:
__del__ is a destructor method that is automatically called when an object is deleted or goes out of scope.
It helps free resources or perform cleanup tasks like closing files or releasing memory.

Q22: What is the difference between @staticmethod and @classmethod in Python?
Answer:

 - @staticmethod: Doesn’t take cls or self, used for independent utility functions.

 - @classmethod: Takes cls as the first argument and can modify class-level attributes.

 Q23: How does polymorphism work in Python with inheritance?

Answer:
Polymorphism in inheritance allows a child class to redefine a parent’s method and still be accessed using a parent reference.
At runtime, Python decides which version to call based on the object type (runtime polymorphism).

Q24: What is method chaining in Python OOP?

Answer:
Method chaining is a technique where multiple methods are called in a single statement by returning self at the end of each method.
This makes code more concise and readable.

Q25: What is the purpose of the __call__ method in Python?

Answer:
The __call__ method allows an object to be called like a function.
If a class defines __call__, its instances can be invoked with parentheses.
It’s often used for function wrappers, decorators, or callable objects.

```python
#Q2: Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle from it and implement the area() method in both.

from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(4, 6)
print(c.area())
print(r.area())
```

```python
#Q3: Implement a multi-level inheritance scenario where a class Vehicle has an attribute type. Derive a class Car and further derive a class ElectricCar that adds a battery attribute.

class Vehicle:
    def __init__(self, type):
        self.type = type

class Car(Vehicle):
    def __init__(self, type, brand):
        super().__init__(type)
        self.brand = brand

class ElectricCar(Car):
    def __init__(self, type, brand, battery):
        super().__init__(type, brand)
        self.battery = battery

e = ElectricCar("Four Wheeler", "Tesla", "100 kWh")
print(e.type, e.brand, e.battery)
```

```python
#Q4: Demonstrate polymorphism by creating a base class Bird with a method fly(). Create two derived classes Sparrow and Penguin that override the fly() method.

class Bird:
    def fly(self):
        print("Some birds can fly")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies high")

class Penguin(Bird):
    def fly(self):
        print("Penguins cannot fly")

for b in [Sparrow(), Penguin()]:
    b.fly()
```

```python
#Q5: Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes balance and methods to deposit, withdraw, and check balance.

class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance
    def deposit(self, amount):
        self.__balance += amount
    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
    def get_balance(self):
        return self.__balance

acc = BankAccount(500)
acc.deposit(200)
acc.withdraw(100)
print(acc.get_balance())
```

```python
#Q6: Demonstrate runtime polymorphism using a method play() in a base class Instrument. Derive classes Guitar and Piano that implement their own version of play().

class Instrument:
    def play(self):
        print("Playing an instrument")

class Guitar(Instrument):
    def play(self):
        print("Playing the guitar")

class Piano(Instrument):
    def play(self):
        print("Playing the piano")

for i in [Guitar(), Piano()]:
    i.play()
```

```python
#Q7: Create a class MathOperations with a class method add_numbers() to add two numbers and a static method subtract_numbers() to subtract two numbers.

class MathOperations:
    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b

print(MathOperations.add_numbers(10, 5))
print(MathOperations.subtract_numbers(10, 5))
```

```python
#Q8: Implement a class Person with a class method to count the total number of persons created.

class Person:
    count = 0
    def __init__(self, name):
        self.name = name
        Person.count += 1

    @classmethod
    def total_persons(cls):
        return cls.count

p1 = Person("Ali")
p2 = Person("Sara")
print(Person.total_persons())
```

```python
#Q9: Write a class Fraction with attributes numerator and denominator. Override the __str__ method to display the fraction as "numerator/denominator".

class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

f = Fraction(3, 4)
print(f)
```

```python
#Q10: Demonstrate operator overloading by creating a class Vector and overriding the __add__ method to add two vectors.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    def __str__(self):
        return f"({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(4, 1)
print(v1 + v2)
```

```python
#Q12: Implement a class Student with attributes name and grades. Create a method average_grade() to compute the average of the grades.

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades
    def average_grade(self):
        return sum(self.grades) / len(self.grades)

s = Student("Saif", [85, 90, 78, 92])
print(s.average_grade())
```

```python
#Q14: Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary.

class Employee:
    def __init__(self, hours, rate):
        self.hours = hours
        self.rate = rate
    def calculate_salary(self):
        return self.hours * self.rate

class Manager(Employee):
    def __init__(self, hours, rate, bonus):
        super().__init__(hours, rate)
        self.bonus = bonus
    def calculate_salary(self):
        return super().calculate_salary() + self.bonus

m = Manager(160, 50, 2000)
print(m.calculate_salary())
```

```python
#Q15: Create a class Product with attributes name, price, and quantity. Implement a method total_price() that calculates the total price of the product.

class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
    def total_price(self):
        return self.price * self.quantity

p = Product("Laptop", 60000, 2)
print(p.total_price())
```

```python
#Q16: Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that implement the sound() method.

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Moo")

class Sheep(Animal):
    def sound(self):
        print("Baa")

for a in [Cow(), Sheep()]:
    a.sound()
```

```python
#Q17: Create a class Book with attributes title, author, and year_published. Add a method get_book_info() that returns a formatted string with the book's details.

class Book:
    def __init__(self, title, author, year_published):
        self.title = title
        self.author = author
        self.year_published = year_published
    def get_book_info(self):
        return f"'{self.title}' by {self.author}, published in {self.year_published}"

b = Book("The Alchemist", "Paulo Coelho", 1988)
print(b.get_book_info())
```

```python
#Q18: Create a class House with attributes address and price. Create a derived class Mansion that adds an attribute number_of_rooms.

class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms

m = Mansion("123 Ocean Drive", 5000000, 10)
print(m.address, m.price, m.number_of_rooms)
```
